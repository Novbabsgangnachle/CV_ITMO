## Задание 1. Случайный массив NumPy и операции над ним

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

### Создание случайного массива

```python
np.random.seed(42)
arr = np.random.randint(0, 100, 20)
print('Одномерный массив:', arr)
```

### Базовые статистические операции

```python
print('Максимум:', arr.max())
print('Минимум:', arr.min())
print('Среднее:', arr.mean())
print('Индекс максимума:', arr.argmax())
print('Индекс минимума:', arr.argmin())
print('Стандартное отклонение:', arr.std())
print('Сумма элементов:', arr.sum())
```

### Reshape — изменение формы массива

```python
mat = arr.reshape(4, 5)
print('Матрица 4x5:')
print(mat)
print('Форма:', mat.shape)
```

### Индексация и срезы

```python
print('Элемент [1, 3]:', mat[1, 3])
print('Первая строка:', mat[0, :])
print('Последний столбец:', mat[:, -1])
print('Подматрица [0:2, 0:3]:')
print(mat[0:2, 0:3])
```

### Арифметические операции

```python
arr2 = np.random.randint(0, 100, 20)
print('arr:', arr)
print('arr2:', arr2)
print()
print('arr + arr2:', arr + arr2)
print('arr * arr2:', arr * arr2)
print('arr * 2:', arr * 2)
print('arr ** 2:', arr ** 2)
```

### Скалярное произведение (dot product)

```python
a = np.random.randn(5)
b = np.random.randn(5)
print('a:', a)
print('b:', b)
print()
print('Поэлементное умножение и сумма:', (a * b).sum())
print('np.dot(a, b):', np.dot(a, b))
print('a.dot(b):', a.dot(b))
```

### Булева индексация (фильтрация)

```python
print('Элементы > 50:', arr[arr > 50])
print('Элементы между 20 и 60:', arr[(arr >= 20) & (arr <= 60)])
```

### Специальные массивы

```python
print('Нулевая матрица 3x3:')
print(np.zeros((3, 3)))
print()
print('Единичная матрица 3x3:')
print(np.ones((3, 3)))
print()
print('Единичная (eye) матрица 4x4:')
print(np.eye(4))
print()
print('arange(0, 20, 3):', np.arange(0, 20, 3))
```

### Транспонирование и сортировка

```python
print('Исходная матрица:')
print(mat)
print()
print('Транспонированная матрица:')
print(mat.T)
print()
print('Отсортированный массив:', np.sort(arr))
```

## Задание 2. Выделение объектов на изображении с помощью OpenCV

```python
import cv2
```

### Загрузка изображения

```python
img_bgr = cv2.imread('DATA/lab_1/1.jpg')
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

print('Размер изображения:', img_rgb.shape)

plt.figure(figsize=(10, 7))
plt.imshow(img_rgb)
plt.title('Исходное изображение')
plt.axis('off')
plt.show()
```

### Выделение объектов функциями рисования OpenCV

На изображении присутствует человек. Выделим его и отдельные элементы (голова, пиджак, брюки, ботинки) с помощью функций `cv2.rectangle`, `cv2.circle`, `cv2.putText`.

```python
img_draw = img_rgb.copy()

# 1. Прямоугольник вокруг всего человека 
cv2.rectangle(img_draw, (120, 30), (940, 1250), (0, 255, 0), 4)

# 2. Круг вокруг головы 
cv2.circle(img_draw, (580, 150), 100, (255, 0, 0), 3)

# 3. Прямоугольник вокруг пиджака 
cv2.rectangle(img_draw, (400, 250), (800, 710), (0, 0, 255), 3)

# 4. Прямоугольник вокруг брюк 
cv2.rectangle(img_draw, (380, 600), (750, 1200), (255, 255, 0), 3)

# 5. Контур левого ботинка 
cv2.rectangle(img_draw, (350, 1140), (510, 1250), (255, 0, 255), 3)

# 6. Контур правого ботинка 
cv2.rectangle(img_draw, (690, 1140), (900, 1250), (255, 0, 255), 3)

font = cv2.FONT_HERSHEY_SIMPLEX
cv2.putText(img_draw, 'Person', (950, 650), font, 1.5, (0, 255, 0), 3, cv2.LINE_AA)
cv2.putText(img_draw, 'Head', (700, 170), font, 1.2, (255, 0, 0), 2, cv2.LINE_AA)
cv2.putText(img_draw, 'Jacket', (810, 500), font, 1.2, (0, 0, 255), 2, cv2.LINE_AA)
cv2.putText(img_draw, 'Pants', (650, 900), font, 1.2, (255, 255, 0), 2, cv2.LINE_AA)
cv2.putText(img_draw, 'Shoe', (250, 1200), font, 1.0, (255, 0, 255), 2, cv2.LINE_AA)
cv2.putText(img_draw, 'Shoe', (590, 1170), font, 1.0, (255, 0, 255), 2, cv2.LINE_AA)

plt.figure(figsize=(10, 7))
plt.imshow(img_draw)
plt.title('Выделение объектов функциями OpenCV')
plt.axis('off')
plt.show()
```

### Автоматическое выделение человека прямоугольником

```python
gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

_, thresh = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY_INV)

contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

largest = max(contours, key=cv2.contourArea)
x, y, w, h = cv2.boundingRect(largest)

img_auto = img_rgb.copy()
cv2.rectangle(img_auto, (x, y), (x + w, y + h), (0, 255, 0), 4)

print(f'Bounding box: x={x}, y={y}, w={w}, h={h}')

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

axes[0].imshow(thresh, cmap='gray')
axes[0].set_title('Бинарная маска (threshold)')
axes[0].axis('off')

axes[1].imshow(img_auto)
axes[1].set_title('Автоматическое выделение человека')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

## Задание 3. Реализация функции rectangle

Реализуем собственную функцию `my_rectangle`, которая рисует прямоугольник на изображении (NumPy-массиве), **не используя OpenCV** для отрисовки.

```python
def my_rectangle(img, pt1, pt2, color=(0, 255, 0), thickness=1):
    result = img.copy()
    h, w = result.shape[:2]
    
    x1, y1 = pt1
    x2, y2 = pt2
    
    # Убеждаемся, что pt1 — верхний левый, pt2 — нижний правый
    x1, x2 = min(x1, x2), max(x1, x2)
    y1, y2 = min(y1, y2), max(y1, y2)
    
    # Обрезаем по границам изображения
    x1 = max(0, x1)
    y1 = max(0, y1)
    x2 = min(w - 1, x2)
    y2 = min(h - 1, y2)
    
    color = np.array(color, dtype=result.dtype)
    
    if thickness == -1:
        # Заливка прямоугольника
        result[y1:y2+1, x1:x2+1] = color
    else:
        t = thickness
        # Верхняя горизонтальная линия
        result[y1:min(y1+t, y2+1), x1:x2+1] = color
        # Нижняя горизонтальная линия
        result[max(y2-t+1, y1):y2+1, x1:x2+1] = color
        # Левая вертикальная линия
        result[y1:y2+1, x1:min(x1+t, x2+1)] = color
        # Правая вертикальная линия
        result[y1:y2+1, max(x2-t+1, x1):x2+1] = color
    
    return result
```

### Тест 1: Прямоугольники на чёрном холсте

```python
canvas = np.zeros((512, 512, 3), dtype=np.uint8)

# Зелёный прямоугольник с толщиной 5
canvas = my_rectangle(canvas, (50, 50), (200, 200), color=(0, 255, 0), thickness=5)

# Красный залитый прямоугольник
canvas = my_rectangle(canvas, (250, 100), (450, 300), color=(255, 0, 0), thickness=-1)

# Синий тонкий прямоугольник
canvas = my_rectangle(canvas, (100, 350), (400, 480), color=(0, 0, 255), thickness=2)

plt.figure(figsize=(6, 6))
plt.imshow(canvas)
plt.title('Прямоугольники на чёрном холсте (my_rectangle)')
plt.axis('off')
plt.show()
```

### Тест 2: Прямоугольник на реальном изображении

```python
test_img_bgr = cv2.imread('DATA/lab_1/1.jpg')
test_img_rgb = cv2.cvtColor(test_img_bgr, cv2.COLOR_BGR2RGB)

result = my_rectangle(test_img_rgb, (100, 20), (950, 1260), color=(0, 255, 0), thickness=5)

plt.figure(figsize=(10, 7))
plt.imshow(result)
plt.title('Прямоугольник на изображении (my_rectangle)')
plt.axis('off')
plt.show()
```

### Тест 3: Сравнение с cv2.rectangle

```python
img_my = my_rectangle(test_img_rgb.copy(), (100, 20), (950, 1260), color=(0, 255, 0), thickness=5)

img_cv = test_img_rgb.copy()
cv2.rectangle(img_cv, (100, 20), (950, 1260), (0, 255, 0), 5)

fig, axes = plt.subplots(1, 2, figsize=(16, 7))

axes[0].imshow(img_my)
axes[0].set_title('my_rectangle (NumPy)')
axes[0].axis('off')

axes[1].imshow(img_cv)
axes[1].set_title('cv2.rectangle (OpenCV)')
axes[1].axis('off')

plt.tight_layout()
plt.show()
```
